```python
# CELL 21: Create CTC Data Collator for MMS ASR

from dataclasses import dataclass
from typing import Dict, List, Union
import torch


@dataclass
class DataCollatorCTCWithPadding:
    processor: object
    padding: Union[bool, str] = True

    def __call__(
        self,
        features: List[Dict[str, Union[List[int], torch.Tensor]]]
    ):

        # Separate audio inputs and labels
        input_features = [
            {"input_values": feature["input_values"]}
            for feature in features
        ]

        label_features = [
            {"input_ids": feature["labels"]}
            for feature in features
        ]

        # Pad audio inputs
        batch = self.processor.feature_extractor.pad(
            input_features,
            padding=self.padding,
            return_tensors="pt"
        )

        # Pad labels
        labels_batch = self.processor.tokenizer.pad(
            label_features,
            padding=self.padding,
            return_tensors="pt"
        )

        # Replace padding tokens with -100
        # so CTC loss ignores them
        labels = labels_batch["input_ids"].masked_fill(
            labels_batch["attention_mask"].ne(1),
            -100
        )

        batch["labels"] = labels

        return batch


# Create the data collator
data_collator = DataCollatorCTCWithPadding(
    processor=english_processor,
    padding=True
)

print("✅ CTC Data Collator created successfully!")

# Test the collator with the current dataset
test_batch = data_collator(
    [prepared_dataset[0]]
)

print("\n📦 Test Batch:")
print("Input shape:", test_batch["input_values"].shape)
print("Labels shape:", test_batch["labels"].shape)

print("\n🏷️ Labels:")
print(test_batch["labels"])
```

```python
# CELL 22: Configure MMS ASR Fine-Tuning

from transformers import TrainingArguments

# Output directory for the fine-tuned model
OUTPUT_DIR = "/content/eldercare_voice_ai/mms_english_model"

# Training configuration
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,

    # Number of training epochs
    num_train_epochs=10,

    # One sample per batch because our dataset is currently very small
    per_device_train_batch_size=1,

    # Accumulate gradients
    gradient_accumulation_steps=1,

    # Learning rate
    learning_rate=1e-5,

    # Save model
    save_strategy="epoch",

    # Logging
    logging_strategy="steps",
    logging_steps=1,

    # Do not evaluate because we have only one sample
    eval_strategy="no",

    # Use GPU if available
    fp16=torch.cuda.is_available(),

    # Keep only the latest checkpoint
    save_total_limit=2,

    # Report nothing to external services
    report_to="none",

    # Disable automatic column removal
    remove_unused_columns=False
)

print("✅ Training configuration created successfully!")

print("\n📋 Training Configuration:")
print("Output directory:", OUTPUT_DIR)
print("Epochs:", training_args.num_train_epochs)
print("Batch size:", training_args.per_device_train_batch_size)
print("Learning rate:", training_args.learning_rate)
print("GPU available:", torch.cuda.is_available())
print("FP16 enabled:", training_args.fp16)
```

```python
# CELL 24: Memory-Safe MMS ASR Fine-Tuning

import torch
import gc

print("🧹 Clearing unused GPU memory...")

# Clear Python garbage
gc.collect()

# Clear CUDA cache
if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("GPU memory cleared.")


# ---------------------------------------------------------
# 1. Freeze the MMS feature extractor
# ---------------------------------------------------------

english_model.freeze_feature_encoder()

print("\n🔒 Feature encoder frozen.")


# ---------------------------------------------------------
# 2. Freeze most of the MMS model
# ---------------------------------------------------------

# Freeze all parameters first
for param in english_model.parameters():
    param.requires_grad = False


# ---------------------------------------------------------
# 3. Enable training only for the CTC classification head
# ---------------------------------------------------------

for param in english_model.lm_head.parameters():
    param.requires_grad = True


print("🎯 Only CTC classification head will be trained.")


# ---------------------------------------------------------
# 4. Count trainable parameters
# ---------------------------------------------------------

total_params = sum(
    p.numel() for p in english_model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in english_model.parameters()
    if p.requires_grad
)

print("\n📊 Model Parameters:")
print("Total parameters:", f"{total_params:,}")
print("Trainable parameters:", f"{trainable_params:,}")

percentage = (trainable_params / total_params) * 100

print(
    "Trainable percentage:",
    round(percentage, 4),
    "%"
)


# ---------------------------------------------------------
# 5. Move model to GPU
# ---------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

english_model = english_model.to(device)

print("\n🚀 Model device:", device)

if torch.cuda.is_available():
    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )


# ---------------------------------------------------------
# 6. Create a new memory-efficient Trainer
# ---------------------------------------------------------

from transformers import Trainer, TrainingArguments

SAFE_OUTPUT_DIR = (
    "/content/eldercare_voice_ai/"
    "mms_english_model"
)

safe_training_args = TrainingArguments(
    output_dir=SAFE_OUTPUT_DIR,

    # Training
    num_train_epochs=10,
    per_device_train_batch_size=1,

    # Learning rate for the trainable CTC head
    learning_rate=1e-4,

    # Memory optimization
    fp16=True,

    # Disable unnecessary features
    eval_strategy="no",
    save_strategy="no",
    logging_strategy="steps",
    logging_steps=1,

    # Important because we already prepared the dataset
    remove_unused_columns=False,

    # No external logging
    report_to="none",

    # Do not keep optimizer checkpoints
    save_only_model=True
)


# ---------------------------------------------------------
# 7. Create Trainer
# ---------------------------------------------------------

trainer = Trainer(
    model=english_model,
    args=safe_training_args,
    train_dataset=prepared_dataset,
    data_collator=data_collator
)

print("\n" + "=" * 60)
print("✅ MEMORY-SAFE TRAINER CREATED")
print("=" * 60)

print("\nTraining samples:", len(prepared_dataset))
print("Epochs:", safe_training_args.num_train_epochs)
print("Batch size:", safe_training_args.per_device_train_batch_size)
print("Learning rate:", safe_training_args.learning_rate)
print("FP16:", safe_training_args.fp16)
print("Trainable parameters:", f"{trainable_params:,}")

# Check available GPU memory
if torch.cuda.is_available():
    free_memory, total_memory = torch.cuda.mem_get_info()

    print("\n💾 GPU Memory:")
    print(
        "Free:",
        round(free_memory / (1024**3), 2),
        "GB"
    )

    print(
        "Total:",
        round(total_memory / (1024**3), 2),
        "GB"
    )

print("\n✅ Ready for memory-safe training.")
```

```python
# CELL 25: Start Memory-Safe MMS ASR Training

import gc
import torch

print("🧹 Preparing GPU memory for training...")

# Python garbage collection
gc.collect()

# Clear unused CUDA memory
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

# Display GPU memory before training
if torch.cuda.is_available():
    free_memory, total_memory = torch.cuda.mem_get_info()

    print(
        "GPU free memory before training:",
        round(free_memory / (1024**3), 2),
        "GB"
    )

    print(
        "GPU total memory:",
        round(total_memory / (1024**3), 2),
        "GB"
    )


print("\n🚀 Starting memory-safe MMS ASR training...")
print("Training samples:", len(prepared_dataset))
print("Epochs:", safe_training_args.num_train_epochs)
print("Trainable parameters:", f"{trainable_params:,}")

print("\n⚠️ This is a 1-sample pipeline demonstration.")
print("Training may overfit this single recording.\n")


# ---------------------------------------------------------
# START TRAINING
# ---------------------------------------------------------

try:

    train_result = trainer.train()

    print("\n" + "=" * 60)
    print("✅ TRAINING COMPLETED SUCCESSFULLY!")
    print("=" * 60)

    print("\n📊 Training Results:")

    if hasattr(train_result, "metrics"):
        for key, value in train_result.metrics.items():
            print(f"{key}: {value}")

except torch.cuda.OutOfMemoryError:

    print("\n❌ CUDA OUT OF MEMORY")
    print("\nThe GPU still does not have enough free memory.")
    print("Do NOT restart the whole project yet.")
    print("We will use an even more memory-efficient approach.")

    # Clear memory
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.ipc_collect()

except Exception as e:

    print("\n❌ Training failed.")
    print("Error type:", type(e).__name__)
    print("Error:", str(e))
```

```python
# CELL 27: Test Fine-Tuned MMS English ASR Model

import torch
import librosa

TEST_AUDIO = (
    "/content/eldercare_voice_ai/"
    "processed_audio/"
    "WhatsApp Ptt 2026-09-05 at 3.26.08 PM.wav"
)

# Put model in evaluation mode
english_model.eval()

# Load audio
audio, sample_rate = librosa.load(
    TEST_AUDIO,
    sr=16000,
    mono=True
)

# Convert audio to model input
inputs = english_processor(
    audio,
    sampling_rate=16000,
    return_tensors="pt"
)

input_values = inputs.input_values.to(device)

# Run inference
with torch.no_grad():
    logits = english_model(input_values).logits

# Get predicted tokens
predicted_ids = torch.argmax(logits, dim=-1)

# Convert tokens to text
prediction = english_processor.batch_decode(
    predicted_ids
)[0]

expected = "It's wake up time"

print("=" * 60)
print("🎤 FINE-TUNED MMS ENGLISH ASR TEST")
print("=" * 60)

print("Expected : ", expected)
print("Predicted: ", prediction)

# Normalize text for comparison
def normalize(text):
    return " ".join(
        text.lower()
        .replace("'", "")
        .split()
    )

if normalize(prediction) == normalize(expected):
    print("\n✅ Prediction matches the training transcription!")
else:
    print("\n⚠️ Prediction differs from the training transcription.")

print("=" * 60)
```
